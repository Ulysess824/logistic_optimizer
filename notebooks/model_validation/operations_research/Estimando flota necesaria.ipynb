# 🚛 Estimación de Flota: Ley de Little y Factor de Utilización

La **Ley de Little** es una herramienta fundamental en logística para determinar cuántos vehículos ($L$) necesitamos para soportar una tasa de demanda constante.

### 1. Flota Teórica (Base)
$$L_{teorica} = \lambda \times W$$

- **$\lambda$ (Lambda)**: Tasa de despachos diarios (camiones/día).
- **$W$ (Wait/Ciclo)**: Tiempo promedio del ciclo logístico (días).

### 2. Flota Real (Con Buffer)
Dado que los activos requieren mantenimiento y existen volatilidades en la demanda, aplicamos un factor de seguridad ($f_{buffer}$):

$$L_{real} = L_{teorica} \times (1 + f_{buffer})$$

Donde para este proyecto se utiliza un buffer estándar del **10%** ($f_{buffer} = 0.10$).
