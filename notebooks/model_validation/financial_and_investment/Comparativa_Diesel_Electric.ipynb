## 📐 Modelado del TCO y Paridad de Costes

Para comparar ambas tecnologías de forma justa, utilizamos el **Coste Total de Propiedad Neto ($TCO_{net}$)**:

$$TCO_{net} = \sum_{t=0}^{n} \frac{OPEX_t + CAPEX_t - TS_t - S_t}{(1+r)^t}$$

Donde:
- $OPEX_t$: Costes operativos (Energía, Mantenimiento).
- $CAPEX_t$: Inversión en activos e infraestructura.
- $TS_t$ (Tax Shield): Escudo fiscal por amortización y gastos.
- $S_t$: Subvenciones directas recibidas.

### Punto de Paridad (Breakeven)
El ahorro operativo del eléctrico ($OPEX_D > OPEX_E$) debe compensar el sobrecoste de inversión ($CAPEX_E > CAPEX_D$):

$$\Delta CAPEX \le \sum_{t=1}^{n} \frac{\Delta OPEX_t}{(1+r)^t}$$
