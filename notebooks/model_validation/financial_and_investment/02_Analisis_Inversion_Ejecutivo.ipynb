# 📈 Tesis de Inversión: Análisis de Flota Mixta (44t)

Este informe técnico evalúa la viabilidad financiera de una flota de **11 camiones** (7 Diésel y 4 Eléctricos) bajo tres modalidades de adquisición. El objetivo es determinar la estructura óptima de capital.

## 📐 Marco Teórico Matemático

### 1. Valor Actual Neto (VAN)
$$VAN = \sum_{t=0}^{n} \frac{CF_t}{(1 + WACC)^t}$$

## ⚙️ 1. Consola de Parámetros de Inversión
Mapeo completo entre los inputs de negocio, las variables de cálculo en Python y su representación en la tesis doctoral.

| Parámetro | Clave (Python) | Variable (LaTeX) | Valor Diésel | Valor Eléctrico |
| :--- | :--- | :---: | :--- | :--- |
| **CAPEX (Inversión)** | `capex` | $C_0$ | 140_000 € | 340_000 € |
| **Consumo Medio** | `consumo_...` | $\eta$ | 33.0 L/100km | 1.30 kWh/km |
| **Precio Energía** | `coste_...` | $P_e$ | 1.45 €/L | 0.18 €/kWh |

| **Seguro Anual** | `seguro_anual_compra` | $S_a$ | 3_500 € | 4_500 € |
| **Valor Residual (5a)** | `residual_value_pct` | $V_R$ | 30% | 25% |
| **Subvención MOVES** | `ayuda_moves` | $A_{moves}$ | N/A | 0 € |

```python
import sys
import os
import pandas as pd
from IPython.display import display, HTML

sys.path.append(os.path.abspath('../../../'))
from logistic_core.utils.investment_analyzer import InvestmentAnalyzer

analyzer = InvestmentAnalyzer()
res = analyzer.generar_tabla_comparativa()

print('✅ Parámetros cargados y motor financiero inicializado.')
```

## 📊 2. Resultados: Matriz de TCO de Flota Mixta
Consolidado de flujos de caja descontados para 11 unidades durante un horizonte de 5 años.

```python
data = []
for mod in ['compra', 'leasing', 'renting']:
    info = res['flota_mixta'][mod]
    data.append({
        'MODALIDAD': mod.upper(),
        '7x Diésel (VAN)': f'{-info["tco_diesel_subtotal"]:,.0f} €'.replace(',', '.'),
        '4x Eléctrico (VAN)': f'{-info["tco_ev_subtotal"]:,.0f} €'.replace(',', '.'),
        'TCO TOTAL FLOTA': f'{-info["tco_total"]:,.0f} €'.replace(',', '.'),
        '€/km TOTAL': f'{info["coste_km_medio"]:.3f} €'
    })

df = pd.DataFrame(data)
display(HTML(df.to_html(index=False)))
```
