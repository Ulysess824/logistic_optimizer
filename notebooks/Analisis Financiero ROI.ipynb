# 📈 Análisis Financiero ROI: Estrategia Asset-Light

Este notebook evalúa la rentabilidad de la inversión bajo el modelo **Asset-Light**. En este escenario, la empresa NO adquiere camiones propios, sino que sigue subcontratando el transporte a terceros, utilizando el software optimizador para reducir los kilómetros recorridos.

### Características del Modelo Asset-Light
- **Baja Inversión (CAPEX)**: Únicamente la licencia del software y consultoría de implementación.
- **Ahorro Directo**: Reducción del gasto mensual en proveedores externos gracias a rutas más eficientes.
- **Flexibilidad**: No se asumen riesgos de mantenimiento, personal ni depreciación de activos pesados.

```python
import sys
import os
import pandas as pd

# Añadir el directorio raíz al path para importar logistic_core
sys.path.append(os.path.abspath(os.path.join('..')))

from logistic_core.utils.financial_analyzer import FinancialAnalyzer
import logistic_core.config as config

# Inicialización del analizador enfocado en Asset-Light
analyzer = FinancialAnalyzer(
    software_capex=config.SOFTWARE_TMS_CAPEX
)
```

## 1. Simulación de Business Case

Evaluamos el ahorro anual proyectado al optimizar las rutas manteniendo el modelo de subcontratación.

```python
km_diarios_actuales = 5000
km_diarios_optimizados = 4200

bc = analyzer.generate_business_case(
    daily_km_baseline=km_diarios_actuales,
    daily_km_optimized=km_diarios_optimizados
)

# Extraemos métricas específicas de Asset-Light
metrics = bc['asset_light']
operational = bc['operational']

summary_data = {
    "KPI": [
        "Kilómetros Ahorrados / Año",
        "Inversión Inicial (Software)",
        "Ahorro Operativo Anual",
        "ROI proyectado (3 años)",
        "Tiempo de Recuperación (Payback)"
    ],
    "Valor": [
        f"{operational['annual_km_saved']:,} km",
        f"{metrics['capex_eur']:,.0f} €",
        f"{metrics['annual_savings_eur']:,.0f} €",
        f"{metrics['roi_3y_pct']:.1f}%",
        f"{metrics['payback_months']:.2f} meses"
    ]
}

df_asset_light = pd.DataFrame(summary_data)
df_asset_light
```

### Conclusión Operativa
Como se observa, la estrategia **Asset-Light** permite obtener un retorno de inversión muy acelerado. Al no tener que amortizar vehículos, el ahorro generado por la optimización del software paga el CAPEX en apenas unos meses.
