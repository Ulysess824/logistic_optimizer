# 🚛 Estimación de Flota: Ley de Little

La **Ley de Little** es una herramienta fundamental en logística para determinar cuántos vehículos ($L$) necesitamos para soportar una tasa de demanda constante.

### Fórmula: $L = \lambda \times W$

- **$L$ (Flota)**: Número total de camiones físicos requeridos.
- **$\lambda$ (Lambda)**: Tasa de despachos diarios (camiones que salen de planta al día).
- **$W$ (Wait/Ciclo)**: Tiempo promedio que un camión tarda en completar su ciclo y volver a estar disponible (en días).

```python
import sys
import os
import math

# Añadir el directorio raíz al path para importar logistic_core
sys.path.append(os.path.abspath(os.path.join('..')))

from logistic_core.utils.fleet_estimator import FleetCapexEstimator
import logistic_core.config as config

# --- 1. Inicialización de la Clase ---
estimator = FleetCapexEstimator(
    daily_dispatch_rate=config.DAILY_TRUCK_OUTBOUND,
    unit_truck_cost=config.CAPEX_TRUCK_UNIT_COST,
    utilization_buffer=config.DEFAULT_FLEET_BUFFER
)

# --- 2. Estimación Basada en Ley de Little ---
w_dias = 1.2 # Ciclo promedio
summary = estimator.generate_investment_summary(average_cycle_time_days=w_dias)

print(f"--- RESULTADOS LEY DE LITTLE (CLASE) ---")
print(f"Lambda (Despachos/día): {summary['daily_dispatch_rate']}")
print(f"W (Ciclo en días): {summary['average_cycle_time_days']}")
print(f"Flota teórica (L): {summary['theoretical_fleet_base']}")
print(f"Flota real necesaria: {summary['final_required_fleet']} camiones")
```
